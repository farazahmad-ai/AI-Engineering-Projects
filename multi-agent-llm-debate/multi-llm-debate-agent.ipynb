```python
import os
import requests
from dotenv import load_dotenv
from openai import OpenAI
from IPython.display import Markdown, display
```

```python
load_dotenv(override=True)
gemini_api_key = os.getenv('GEMINI_API_KEY')
openrouter_api_key = os.getenv('OPENROUTER_API_KEY')
```

```python
gemini_url = "https://generativelanguage.googleapis.com/v1beta/openai/"
openrouter_url = "https://openrouter.ai/api/v1"
ollama_url = "http://localhost:11434/v1"


gemini = OpenAI(api_key=gemini_api_key, base_url=gemini_url)
openrouter = OpenAI(base_url=openrouter_url, api_key=openrouter_api_key)
ollama = OpenAI(api_key="ollama", base_url=ollama_url)
```

```python
gpt_model = "gpt-4.1-nano"
claude_model = "claude-3-haiku"
gemini_model = "google/gemini-2.5-flash"


name_gpt = "Arthur (Platonist Mathematician)"
name_claude = "Formalist Mathematician"
name_gemini = "Soren (Modern Philosopher of Cognitive Science)"
```

```python
arthur_system = """You are Arthur, a Platonist Mathematician. 
You firmly believe that mathematical truths, numbers, and geometric structures exist objectively 
in an abstract reality independently of humans. Humans do NOT invent math; we DISCOVER it, 
just as explorers discover uncharted continents."""

armstrong_system = """You are Armstrong, a Formalist Mathematician. 
You firmly believe that mathematics is a HUMAN INVENTION—a game of logical symbols, axioms, 
and formal rules created by human minds (like chess) to organize patterns and solve problems."""

soren_system = """You are Soren, a Modern Philosopher of Cognitive Science and Moderator. 
You synthesize the debate between Platonism and Formalism. You introduce ideas like embodied cognition 
(how human brain evolution shapes math) and challenge both debaters to avoid dogmatism."""
```

```python
transcript = """Arthur (Platonist): Look at prime numbers or the 
Mandelbrot set—they possessed their unique mathematical properties 
billions of years before humans evolved. We didn't invent the ratio Pi; 
it is built into the geometry of spacetime. We discovered it!

Armstrong (Formalist): That's a romantic illusion, Arthur. We invented 
the concept of 'circles', we invented the base-10 number system, and we 
invented the axioms of Euclidean geometry. If intelligent alien life 
exists, their mathematical syntax and axiomatic systems would look 
entirely different from ours.

Soren (Philosopher): Both of you highlight the central paradox. 
Arthur sees the uncanny effectiveness of math in physics as proof of 
objective existence, while Armstrong sees math as a specialized human 
cognitive language. Let's dig deeper: how do we explain why an 
'invented' system predicts physical reality so accurately?"""
```

```python
def call_agent(model_name, system_prompt, agent_name, current_transcript):

    user_prompt = f"""The debate so far between Arthur, Armstrong, and Soren is:
{current_transcript}

Now, respond with your next turn as {agent_name}. 
Speak strictly as {agent_name} in 2-3 sentences. Maintain your specific philosophy 
and address the other debaters directly. Do not add any name 
prefixes like '{agent_name}:'."""
    
    messages = [
        {"role": "system", "content": system_prompt},
        {"role": "user", "content": user_prompt}
    ]
    
    response = openrouter.chat.completions.create(
        model=model_name,
        messages=messages
    )
    return response.choices[0].message.content.strip()
```

```python

for round_num in range(1, 3):
    print(f"\n=================== ROUND {round_num} ===================")
    
   
    reply_arthur = call_agent(gpt_model, arthur_system, "Arthur (Platonist)", transcript)
    transcript += f"\n{name_gpt}: {reply_arthur}"
    print(f"\n[1] {name_gpt}:\n{reply_arthur}")
    
 
    reply_armstrong = call_agent(claude_model, armstrong_system, name_claude, transcript)
    transcript += f"\n{name_claude}: {reply_armstrong}"
    print(f"\n[2] {name_claude}:\n{reply_armstrong}")
    
 
    reply_soren = call_agent(gemini_model, soren_system, name_gemini, transcript)
    transcript += f"\n{name_gemini}: {reply_soren}"
    print(f"\n[3] {name_gemini}:\n{reply_soren}")
```
